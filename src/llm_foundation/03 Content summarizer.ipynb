```python
import os 
from dotenv import load_dotenv
from bs4 import BeautifulSoup
import requests
from openai import OpenAI
from IPython.display import display, Markdown
```

```python
load_dotenv(override=True)
MODEL_NAME= os.getenv("MODEL_NAME")
BASE_URL = os.getenv("OLLAMA_HOST")
API_KEY = os.getenv("OLLAMA_API_KEY")
```

```python
# funcation for fetching details from URL 

url = "https://quotes.toscrape.com/"

def wesite_fetcher(url) -> str:
    response = requests.get(url)
    soup = BeautifulSoup(response.content,"html.parser")
    title = soup.title.text if soup.title.text else None
    for irrelevantTag in soup.body(['script',"img","style","nav","footer"]):
        irrelevantTag.decompose()
    
    text = soup.get_text(separator="\n",strip=True)
    return title + " " + text
```

```python
user_prefix_msg= """
Here is the content of the website - please symarize the same in markdown
"""

def message_generator(website):
    return [
        {"role":"system", "content":"you are a helpful assistent"},
        {"role":"user","content":user_prefix_msg + website}
    ]
    
```

```python
openai = OpenAI(
    api_key = API_KEY,
    base_url = BASE_URL
)

def summarize_url(url):
    website = wesite_fetcher(url)
    response = openai.chat.completions.create(model = MODEL_NAME,messages = message_generator(website))
    return response.choices[0].message.content , response.json()
```

```python
def display_summary(url):
    summary  = summarize_url(url)
    display(Markdown(summary))

display_summary(url)
```
