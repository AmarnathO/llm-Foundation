```python
from openai import OpenAI
from dotenv import load_dotenv
import os 
from IPython.display import display, Markdown
```

```python
OLLAMA_HOST = os.getenv("OLLAMA_HOST")
API_KEY = os.getenv("OLLAMA_API_KEY")
MODEL_NAME = os.getenv("MODEL_NAME")
```

```python
openai = OpenAI(
    api_key=API_KEY,
    base_url= OLLAMA_HOST
)

msg = "Hey there my name is Amar"
system_msg = "You are a snarky assistant"

message = [
    {"role":"system","content":system_msg},
    {"role":"user","content":msg}
]

response = openai.chat.completions.create(model = MODEL_NAME, temperature=0.9,messages=message)
display(Markdown(response.choices[0].message.content))
```

```python
message_1 = [
    {"role":"user","content":"What's is my name"}
    ]

response = openai.chat.completions.create(model=MODEL_NAME,temperature=0.6,messages=message_1)
display(Markdown(response.choices[0].message.content))
```

- Every call to LLM is stateless, It don;t remember the past details 
- We pass in the whole conversastion everytime with prompt that give us illusion of llm have memory 


```python
message_2 = [
    {"role":"system","content":system_msg},
    {"role":"user","content":msg},
    {"role":"Assistant","content": "hey Amar how are you - I believe you like to play cricket"},
    {"role":"user","content":" Which game i like to play"}
]

response = openai.chat.completions.create(model=MODEL_NAME,temperature=0.6,messages = message_2)
display(Markdown(response.choices[0].message.content))
```
