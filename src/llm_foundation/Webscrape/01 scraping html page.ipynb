```python
from bs4 import BeautifulSoup
```

```python
html = """
<html>
  <body>
    <h1>Welcome to My Store</h1>
    <p class="description">Best electronics store</p>
    <a href="/products">View Products</a>
  </body>
</html>
"""
```

```python
soup = BeautifulSoup(html,"html.parser")
print(soup.h1.text)
print(soup.p.text)
print(soup.a['href'])
```

**Most important Beautiful Soup method**

These six methods cover the majority of everyday parsing tasks.

**Method**	            **Purpose**\
find()	        : Find the first matching element \
find_all()	    : Find all matching elements\
select_one()    : 	Find first element using a CSS selector\
select()	    : Find all elements using CSS selectors\
get_text()	    : Extract text\
get()	        : Extract an attribute safely

```python
html = """
<div class="product">
    <h2 class="title">iPhone</h2>
    <span class="price">79900</span>
</div>

<div class="product">
    <h2 class="title">Samsung</h2>
    <span class="price">69900</span>
</div>
"""

soup = BeautifulSoup(html,"html.parser")
products = soup.find_all("div",class_="product")

for product in products:
    name = product.find("h2",class_="title").get_text(strip=True)
    price = product.find("span",class_="price").get_text(strip=True)

    print(name,price)
```
