```python
import os 
from openai import OpenAI
from dotenv import load_dotenv
```

```python
load_dotenv(override=True)

MODEL_Name = os.getenv("MODEL_NAME")
OLLAMA_BASE_URL = os.getenv("OLLAMA_HOST")
OLLAMA_API_KEY = os.getenv("OLLAMA_API_KEY")

openai = OpenAI(api_key=OLLAMA_API_KEY,
base_url = OLLAMA_BASE_URL)
```

```python
user_message = "Hey Tell me about yourself how you come to this planet"
system_message = "You are witty girl you always repsonse with a taunting answer"

messages = [
    {"role":"system","content":system_message},
    {"role":"user","content":user_message}
]

response = openai.chat.completions.create(model=MODEL_Name,messages = messages,temperature=0.8)
print(response.choices[0].message.content)
```

```python
# costing 
print(f"input tokens:{response.usage.prompt_tokens}")
print(f"output tokens:{response.usage.completion_tokens}")
print(f"Total tokens:{response.usage.total_tokens}")
#print(f"Total Cost: {response_hidden_params['response_cost'] * 100:.4f} cents")
```
