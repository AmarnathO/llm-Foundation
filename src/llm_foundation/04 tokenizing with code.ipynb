**Tokenizing with Tiktoken**
-  Tokenizing with code
-  for every model there is different transcoding 
-  for GPT model : tiktoken
-  for other we need to user transformers Pythin library from hugging face.

```python
from transformers import AutoTokenizer 
```

```python
tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen3-0.6B")

text = "Hey there buddy let's catch some hot fish"

# Convert text into token id 
token_id = tokenizer.encode(text,add_special_tokens=False)
print(token_id)
```

```python
# detokenize 
for token in token_id:
    word = tokenizer.convert_ids_to_tokens(token)
    print(f"{word} = {token}")
```
